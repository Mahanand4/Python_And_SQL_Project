```python
# Delivery performance analysis

delivered_orders = orders[
orders['order_delivered_customer_date'].notna()
].copy()

delivered_orders['delivery_days'] = (
delivered_orders['order_delivered_customer_date']
- delivered_orders['order_purchase_timestamp']
).dt.days

print("Average delivery time:",
round(delivered_orders['delivery_days'].mean(), 2), "days")

print("Minimum delivery time:",
delivered_orders['delivery_days'].min(), "days")

print("Maximum delivery time:",
delivered_orders['delivery_days'].max(), "days")
```

```python
# Actual vs estimated delivery performance

delivered_orders['delivery_delay_days'] = (
delivered_orders['order_delivered_customer_date']
- delivered_orders['order_estimated_delivery_date']
).dt.days

print("Average delivery delay:",
round(delivered_orders['delivery_delay_days'].mean(), 2), "days")

print("Orders delivered early/on time:",
(delivered_orders['delivery_delay_days'] <= 0).sum())

print("Orders delivered late:",
(delivered_orders['delivery_delay_days'] > 0).sum())
```

```python
# Delivery performance percentage

total_delivered = len(delivered_orders)

early_on_time_pct = (
(delivered_orders['delivery_delay_days'] <= 0).sum()
/ total_delivered * 100
)

late_pct = (
(delivered_orders['delivery_delay_days'] > 0).sum()
/ total_delivered * 100
)

print("Early/On-time delivery rate:",
round(early_on_time_pct, 2), "%")

print("Late delivery rate:",
round(late_pct, 2), "%")
```

```python
# RFM Analysis

# Reference date
reference_date = orders['order_purchase_timestamp'].max() + pd.Timedelta(days=1)

# Create RFM metrics
rfm = orders[orders['order_status'] == 'delivered'].groupby('customer_id').agg(
Recency=('order_purchase_timestamp',
lambda x: (reference_date - x.max()).days),
Frequency=('order_id', 'nunique')
).reset_index()

# Calculate Monetary value
customer_revenue = (
order_items.groupby('order_id')['price']
.sum()
.reset_index()
.rename(columns={'price': 'order_revenue'})
)

delivered_orders = orders[orders['order_status'] == 'delivered']

rfm = rfm.merge(
delivered_orders[['order_id', 'customer_id']],
on='customer_id',
how='left'
)

rfm = rfm.merge(
customer_revenue,
on='order_id',
how='left'
)

rfm = (
rfm.groupby(['customer_id', 'Recency', 'Frequency'])['order_revenue']
.sum()
.reset_index()
.rename(columns={'order_revenue': 'Monetary'})
)

print("RFM Analysis:")
print(rfm.head())
```

```python
# RFM Scoring

rfm['R_score'] = pd.qcut(
rfm['Recency'].rank(method='first'),
5,
labels=[5, 4, 3, 2, 1]
)

rfm['F_score'] = pd.qcut(
rfm['Frequency'].rank(method='first'),
5,
labels=[1, 2, 3, 4, 5]
)

rfm['M_score'] = pd.qcut(
rfm['Monetary'].rank(method='first'),
5,
labels=[1, 2, 3, 4, 5]
)

rfm['RFM_score'] = (
rfm['R_score'].astype(str) +
rfm['F_score'].astype(str) +
rfm['M_score'].astype(str)
)

print("RFM Scoring:")
print(rfm.head(10))
```

```python
# RFM Customer Segmentation

rfm['Customer_Segment'] = 'Needs Attention'

rfm.loc[
(rfm['R_score'] >= 4) &
(rfm['F_score'] >= 4) &
(rfm['M_score'] >= 4),
'Customer_Segment'
] = 'Champions'

rfm.loc[
(rfm['R_score'] >= 3) &
(rfm['F_score'] >= 3) &
(rfm['Customer_Segment'] == 'Needs Attention'),
'Customer_Segment'
] = 'Loyal Customers'

rfm.loc[
(rfm['R_score'] >= 4) &
(rfm['F_score'] <= 2) &
(rfm['Customer_Segment'] == 'Needs Attention'),
'Customer_Segment'
] = 'New/Promising Customers'

rfm.loc[
(rfm['R_score'] <= 2) &
(rfm['F_score'] >= 3) &
(rfm['Customer_Segment'] == 'Needs Attention'),
'Customer_Segment'
] = 'At Risk Customers'

print("RFM Customer Segmentation:")
print(rfm['Customer_Segment'].value_counts())
```
